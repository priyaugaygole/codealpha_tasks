```python
import pandas as pd

df = pd.read_csv("Students.csv")

print("Rows and Columns:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# Basic information about the dataset

print("Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# AI Tools Usage
plt.figure(figsize=(10,5))
sns.countplot(data=df, y="AI_Tools_Used", order=df["AI_Tools_Used"].value_counts().index)

plt.title("Most Used AI Tools Among Students")
plt.xlabel("Number of Students")
plt.ylabel("AI Tool")
plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.scatterplot(
    data=df,
    x="Daily_Usage_Hours",
    y="Impact_on_Grades"
)

plt.title("AI Usage Hours vs Impact on Grades")
plt.xlabel("Daily AI Usage (Hours)")
plt.ylabel("Impact on Grades")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.boxplot(
    data=df,
    x="Awareness_Level",
    y="Trust_in_AI_Tools"
)

plt.title("AI Awareness Level vs Trust in AI Tools")
plt.xlabel("Awareness Level")
plt.ylabel("Trust in AI Tools")
plt.grid(axis="y")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

numeric_cols = [
    'Year_of_Study',
    'Daily_Usage_Hours',
    'Trust_in_AI_Tools',
    'Impact_on_Grades',
    'Awareness_Level'
]

plt.figure(figsize=(8,6))
sns.heatmap(
    df[numeric_cols].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Between Student AI Usage Factors")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Year_of_Study',
    y='Daily_Usage_Hours'
)

plt.title("AI Usage Hours by Year of Study")
plt.xlabel("Year of Study")
plt.ylabel("Daily AI Usage Hours")
plt.show()
```

```python
# Key Insights from EDA

print("KEY INSIGHTS")
print("=" * 50)

print("1. ChatGPT is the most frequently used AI tool among students.")
print("2. Students use AI tools for different academic purposes such as assignments, coding and learning.")
print("3. Daily AI usage varies across students and study years.")
print("4. The correlation between Daily AI Usage Hours and Impact on Grades is weak.")
print("5. Trust in AI tools varies across different awareness levels.")
print("6. The dataset contains 3614 student records and 16 features.")
```

```python
# Final Summary

print("FINAL SUMMARY")
print("=" * 50)

print(f"Total Students: {df.shape[0]}")
print(f"Total Features: {df.shape[1]}")
print(f"Most Used AI Tool: {df['AI_Tools_Used'].value_counts().idxmax()}")
print(f"Average Daily AI Usage: {df['Daily_Usage_Hours'].mean():.2f} hours")
print(f"Average Trust in AI Tools: {df['Trust_in_AI_Tools'].mean():.2f}/5")
print(f"Average Awareness Level: {df['Awareness_Level'].mean():.2f}/10")
```

```python
# Sentiment Analysis on Use_Cases

from textblob import TextBlob
import matplotlib.pyplot as plt

# Calculate sentiment polarity
df["Sentiment_Score"] = df["Use_Cases"].apply(
    lambda x: TextBlob(str(x)).sentiment.polarity
)

# Convert score into sentiment category
def get_sentiment(score):
    if score > 0:
        return "Positive"
    elif score < 0:
        return "Negative"
    else:
        return "Neutral"

df["Sentiment"] = df["Sentiment_Score"].apply(get_sentiment)

# Show results
print("Sentiment Distribution:")
print(df["Sentiment"].value_counts())

# Plot
df["Sentiment"].value_counts().plot(kind="bar")

plt.title("Sentiment Analysis of Student AI Use Cases")
plt.xlabel("Sentiment")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.show()
```

```python
# Final Sentiment Summary

print("SENTIMENT ANALYSIS SUMMARY")
print("=" * 40)

sentiment_counts = df["Sentiment"].value_counts()

print(sentiment_counts)

print("\nSentiment Percentages:")
print((sentiment_counts / len(df) * 100).round(2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

folder = "/content/AI_Sentiment_Data/Aspect-Based Sentiment Analysis on Generative AI"

file_path = folder + "/Aspect-Based Sentiment Analysis on Generative AI.csv"

sentiment_df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", sentiment_df.shape)
print("\nSentiment Distribution:")
print(sentiment_df["sentiment"].value_counts())

print("\nSample Reviews:")
display(sentiment_df[["review_text", "sentiment"]].head(10))
```

```python
import matplotlib.pyplot as plt

sentiment_counts = sentiment_df["sentiment"].value_counts()

plt.figure(figsize=(8,5))
sentiment_counts.plot(kind="bar")

plt.title("Sentiment Distribution of Generative AI Reviews")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
tool_sentiment = pd.crosstab(
    sentiment_df["ai_tools"],
    sentiment_df["sentiment"]
)

print(tool_sentiment)

tool_sentiment.plot(
    kind="bar",
    figsize=(10,6)
)

plt.title("Sentiment Distribution by AI Tool")
plt.xlabel("AI Tool")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Sentiment")

plt.tight_layout()
plt.show()
```

```python
# AI Tool-wise Sentiment Analysis

tool_sentiment = pd.crosstab(
    sentiment_df["ai_tools"],
    sentiment_df["sentiment"]
)

print(tool_sentiment)

tool_sentiment.plot(kind="bar", figsize=(10,6))

plt.title("Sentiment Distribution by AI Tool")
plt.xlabel("AI Tool")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Sentiment")

plt.tight_layout()
plt.show()
```

```python
aspect_sentiment = pd.crosstab(
    sentiment_df["aspect"],
    sentiment_df["sentiment"]
)

print(aspect_sentiment)

aspect_sentiment.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Sentiment Distribution by AI Aspect")
plt.xlabel("Aspect")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Sentiment")

plt.tight_layout()
plt.show()
```
